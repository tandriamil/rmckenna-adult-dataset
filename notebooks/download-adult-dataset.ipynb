# Download the adult dataset

This notebook downloads the
[adult dataset](https://archive.ics.uci.edu/ml/datasets/adult) using the
[ucimlrepo](github.com/uci-ml-repo/ucimlrepo) tool. We do this to avoid
downloading it on each call to `fetch_ucirepo`.

```python
from pathlib import Path
from ucimlrepo import fetch_ucirepo

ADULT_DATASET_ID = 2

DATA_PATH = Path('../data')
assert DATA_PATH.is_dir() 
```

## Download the dataset

```python
adult = fetch_ucirepo(id=ADULT_DATASET_ID)
```

## Limit to the features

The dataset comes with the metadata and the target variables to predict. For
our use case, **we only use the features**. 

```python
dataset = adult.data.features
dataset.sample(10)
```

## Save the dataset

```python
dataset_path = DATA_PATH / Path('adult-dataset-pickle.zip')
dataset.to_pickle(dataset_path)
```
